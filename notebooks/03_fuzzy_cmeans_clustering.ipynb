```python
import os
import torch

# Paths where you saved files
models_path = "D:/Models"
latents_path = "D:/LatentFeatures"

# Check models
print("=== Saved Models ===")
for file in os.listdir(models_path):
    if file.endswith(".pth"):
        print(file, "->", os.path.getsize(os.path.join(models_path, file)), "bytes")
        # Try loading one to confirm it's valid
        state_dict = torch.load(os.path.join(models_path, file))
        print("  Keys in state_dict:", list(state_dict.keys())[:5], "...")

# Check latent features
print("\n=== Saved Latent Features ===")
for root, dirs, files in os.walk(latents_path):
    for file in files:
        print(os.path.join(root, file), "->", os.path.getsize(os.path.join(root, file)), "bytes")
```

```python
import glob
import numpy as np

latent_files = glob.glob("D:/LatentFeatures/*_train/*.npy")

latents = [np.load(f) for f in latent_files]

latents = np.array(latents)

print("Latent shape:", latents.shape)
```

```python
import skfuzzy as fuzz

X = latents.T   # (features, samples) → (30, 5042)

cntr, u, u0, d, jm, p, fpc = fuzz.cluster.cmeans(
    X, c=3, m=2, error=0.005, maxiter=1000
)

print("FPC:", fpc)
print("Membership shape:", u.shape)
print("Centroids shape:", cntr.shape)
```

```python
np.save("autoencoder_latent_features.npy", latents)
np.save("fcm_membership_matrix.npy", u.T)   # (5042, 3)
np.save("fcm_cluster_centers.npy", cntr)

print("✅ All files saved successfully")
```
